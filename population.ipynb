```python
import random
import statistics


# PART 1 — POPULATION GENERATOR


def generate_population(N):
    """
    Generates a full virtual school population of N students.
    Each student is represented as a dictionary with:
      - study_hours (0–6)
      - sleep_hours (4–10)
      - gaming_hours (0–8)
      - social_media_hours (0–6)
      - grade_level (9–12)
    """
    population = []

    for _ in range(N):
        student = {
            "study_hours": random.uniform(0, 6),
            "sleep_hours": random.uniform(4, 10),
            "gaming_hours": random.uniform(0, 8),
            "social_media_hours": random.uniform(0, 6),
            "grade_level": random.choice([9, 10, 11, 12])
        }
        population.append(student)

    return population


def population_mean(population, column):
    """Returns the true population mean for a given column."""
    return statistics.mean([row[column] for row in population])


# ============================================================
# PART 2 — RANDOM SAMPLING
# ============================================================

def take_random_sample(population, n):
    """
    Takes a random sample of n students from the population.
    Ensures no student is selected twice.
    """
    return random.sample(population, n)


def sample_mean(sample, column):
    """Computes the mean of a given column for a sample."""
    return statistics.mean([row[column] for row in sample])


# ============================================================
# PART 3 — BIASED SAMPLING
# ============================================================

def take_biased_sample(population, n):
    """
    Creates a biased sample by selecting only students
    who game more than 5 hours per day.
    """
    filtered = [row for row in population if row["gaming_hours"] > 5]

    # Ensure we have enough students to sample from
    if len(filtered) < n:
        raise ValueError("Not enough students meet the bias condition.")

    return random.sample(filtered, n)


# ============================================================
# PART 4 — SAMPLE SIZE EXPERIMENT
# ============================================================

def run_experiment(population, column, n):
    """
    Runs 10 random samples of size n and returns the list of sample means.
    """
    means = []
    for _ in range(10):
        sample = take_random_sample(population, n)
        means.append(sample_mean(sample, column))
    return means


# ============================================================
# MAIN PROGRAM EXECUTION
# ============================================================

# ---- Generate population ----
N = 400
population = generate_population(N)

# Choose main column for analysis
column = "gaming_hours"

# ---- Compute true population mean ----
true_mean = population_mean(population, column)
print("====================================")
print("TRUE POPULATION MEAN:", true_mean)
print("====================================\n")

# ---- Part 2: Random sampling repeated 10 times ----
print("RANDOM SAMPLE MEANS (n = 30):")
sample_means = []
for i in range(10):
    sample = take_random_sample(population, 30)
    m = sample_mean(sample, column)
    sample_means.append(m)
    print(f"Sample {i+1} mean: {m}")

print("\nAverage of sample means:", statistics.mean(sample_means))
print("Difference from population mean:", statistics.mean(sample_means) - true_mean)

# ---- Part 3: Biased sample ----
print("\n====================================")
print("BIASED SAMPLE RESULTS")
print("====================================")

biased_sample = take_biased_sample(population, 30)
biased_mean = sample_mean(biased_sample, column)

print("Biased sample mean:", biased_mean)
print("Random sample mean (avg):", statistics.mean(sample_means))
print("Population mean:", true_mean)

# ---- Part 4: Sample size experiment ----
print("\n====================================")
print("SAMPLE SIZE EXPERIMENT")
print("====================================")

sizes = [5, 10, 30, 100]

for n in sizes:
    means = run_experiment(population, column, n)
    print(f"\nSample size {n}:")
    print("Means:", means)
    print("Average of means:", statistics.mean(means))
    print("Difference from population:", statistics.mean(means) - true_mean)
```
